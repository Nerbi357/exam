```python
import os
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from rink import plot_rink

pd.set_option("display.width", 200)
pd.set_option("display.max_columns", 50)
pd.set_option("display.max_colwidth", 80)
os.makedirs("figures", exist_ok=True)
sns.set_theme(style="whitegrid")

DATA = "NHL-playoffs-2026.xlsx"
SHOT_TYPES = ["shot-on-goal", "goal", "missed-shot"]       # броски, не заблокированные
ALL_ATTEMPTS = SHOT_TYPES + ["blocked-shot"]               # все броски
ON_TARGET = ["shot-on-goal", "goal"]                       # точные броски


def header(n, text):
    print("\n" + "=" * 100)
    print(f"Пункт {n}. {text}")
    print("=" * 100)
```

### Пункт 1

```python
header(1, "Чтение данных")
events = pd.read_excel(DATA, index_col=0)    # первый столбец файла — просто номер строки
print("events.shape =", events.shape)
print(events.iloc[:10, [2, 3, 4, 9, 12]])
```

### Пункт 2

```python
header(2, "Сводная таблица по играм")
last = events.sort_values(["Game ID", "Period", "Play ID"]).groupby("Game ID").tail(1).set_index("Game ID")
assert (last["Play Type"] == "game-end").all()
periods = events.groupby("Game ID")["Period"].max()


def outcome(p):
    if p <= 3:
        return "Final"
    return "Final/OT" if p == 4 else f"Final/{p - 3}OT"


games = pd.DataFrame({
    "Date": last["Date"],
    "Away Team": last["Away Team"],
    "Home Team": last["Home Team"],
    "Away Score": last["Cumulated Away Score"],
    "Home Score": last["Cumulated Home Score"],
    "Outcome": periods.map(outcome),
})
# контроль: счёт из game-end совпадает с числом событий goal
goals_cnt = events[events["Play Type"] == "goal"].groupby(["Game ID", "Team Making the Play"]).size()
chk_home = games.apply(lambda r: goals_cnt.get((r.name, r["Home Team"]), 0), axis=1)
chk_away = games.apply(lambda r: goals_cnt.get((r.name, r["Away Team"]), 0), axis=1)
print("Счёт из game-end совпадает с числом событий goal:",
      bool((chk_home == games["Home Score"]).all() and (chk_away == games["Away Score"]).all()))
print(games.columns)
print(games.iloc[-5:, -4:])
print(games["Outcome"].value_counts())
```

### Пункт 3

```python
header(3, "Функция gameInfo")


def gameInfo(gameId):
    g = games.loc[gameId]
    return f"{g['Home Team']} - {g['Away Team']} ({g['Home Score']} - {g['Away Score']}, {g['Outcome']}, {g['Date']})"


print(gameInfo(games.index.min()))
print(gameInfo(games.index.max()))
```

### Пункт 4

```python
header(4, "Число голов в игре")
games["Total Score"] = games["Away Score"] + games["Home Score"]
print(games["Total Score"].describe())
```

### Пункт 5

```python
header(5, "Гистограмма числа голов")
plt.figure(figsize=(7, 4))
sns.countplot(x="Total Score", data=games, color="steelblue",
              order=range(games["Total Score"].min(), games["Total Score"].max() + 1))
plt.xlabel("Число голов")
plt.ylabel("Число матчей")
plt.title("Число голов в одном матче")
plt.tight_layout()
plt.savefig("figures/fig1_total_goals.png", dpi=150)
plt.show()
print(games["Total Score"].value_counts().sort_index().to_string())
print("Доля матчей с 4..7 голами: %.3f" % games["Total Score"].between(4, 7).mean())
```

### Пункт 6

```python
header(6, "Самые нерезультативные и самые результативные игры")
mn, mx = games["Total Score"].min(), games["Total Score"].max()
print("min =", mn, ":", [gameInfo(i) for i in games.index[games["Total Score"] == mn]])
print("max =", mx, ":", [gameInfo(i) for i in games.index[games["Total Score"] == mx]])
```

### Пункт 7

```python
header(7, "Число участвовавших игроков")
away_cols = [c for c in events.columns if c.startswith("Away Team's Player on the Ice")] + ["Away Team's Goalie"]
home_cols = [c for c in events.columns if c.startswith("Home Team's Player on the Ice")] + ["Home Team's Goalie"]
pl_cols = away_cols + home_cols


def players_in_game(d):
    return pd.unique(d[pl_cols].values.ravel())


pcount = events.groupby("Game ID").apply(lambda d: len({p for p in players_in_game(d) if isinstance(p, str)}))  # только настоящие имена
pcount = pcount.rename("Player Count")
top = pcount[pcount == pcount.max()]
res = pd.DataFrame({"Game Info": [gameInfo(i) for i in top.index], "Player Count": top.values}, index=top.index)
print(res.to_string())
print(pcount.describe())
# замечание: pd.unique по всем столбцам учитывает и пропуск (NaN) как «значение»; в игре 2025030216 вратарь
# какой-то команды в некоторые моменты отсутствует на льду, поэтому «наивный» подсчёт даёт 41 вместо 40 игроков
_d = events[events["Game ID"] == 2025030216]
print("Контроль для игры 2025030216: уникальных значений вместе с NaN =", len(pd.unique(_d[pl_cols].values.ravel())),
      "; настоящих игроков =", len(players_in_game(_d)) - 1)
```

### Пункт 8

```python
header(8, "Расстояние до ворот")
GOAL_HALF = 3


def dist(x, y, X):
    dy = np.maximum(np.abs(y) - GOAL_HALF, 0)       # 0 для -3 <= y <= 3
    return np.sqrt((x - X) ** 2 + dy ** 2)


is_shot = events["Play Type"].isin(SHOT_TYPES)
home_shot = events["Team Making the Play"] == events["Home Team"]
left = events["Home Team's Defending Side"] == "left"
# хозяева, защищающие левую сторону, бросают в правые ворота (X=89); гости — в левые (X=-89)
target_x = np.where(home_shot == left, 89, -89)
events["Shot Length"] = np.where(is_shot,
                                 dist(events["X Coordinate on the Ice Rink"],
                                      events["Y Coordinate on the Ice Rink"], target_x),
                                 np.nan)
print(events.loc[is_shot, ["Game ID", "Play Type", "Shot Length"]])
print(events.loc[is_shot, "Shot Length"].describe())
```

### Пункт 9

```python
header(9, "Распределение длины точных и неточных бросков")
sh = events[is_shot].copy()
sh["Shot"] = np.where(sh["Play Type"] == "missed-shot", "missed", "on goal")
g = sns.displot(data=sh, x="Shot Length", hue="Shot", stat="density", common_norm=False,
                bins=range(0, 200, 5), height=4, aspect=1.6, hue_order=["missed", "on goal"])
g.set_axis_labels("Shot length, feet", "Frequency")
plt.savefig("figures/fig2_shot_length.png", dpi=150)
plt.show()
print(sh.groupby("Shot")["Shot Length"].describe())
print("Квантили:\n", sh.groupby("Shot")["Shot Length"].quantile([.5, .9, .99]).unstack())
far = sh[(sh["Shot"] == "on goal") & (sh["Shot Length"] > 100)].copy()
far_home = far["Team Making the Play"] == far["Home Team"]
far["Opp Goalie"] = np.where(far_home, far["Away Team's Goalie"], far["Home Team's Goalie"])
print("Точных бросков дальше 100 футов:", len(far),
      "; из них голов:", (far["Play Type"] == "goal").sum(),
      "; в пустые ворота (вратарь соперника не на льду):", far["Opp Goalie"].isna().sum())
print(far["Play Type"].value_counts().to_string())
# для сравнения: доля пустых ворот по всем точным броскам и все броски с расстояния > 100
print("Всего бросков с расстояния >100 футов (включая промахи):", int((sh["Shot Length"] > 100).sum()))
far["Late P3"] = (far["Period"] >= 3) & (far["Remaining Time"] < "05:00")
print("Зона, в которой находился бросавший (Play Zone) для далёких точных бросков:")
print(far["Play Zone"].value_counts().to_string())
print("Далёкие голы:", int((far["Play Type"] == "goal").sum()), "; из них в последние 5 минут 3-го периода или позже:",
      int(((far["Play Type"] == "goal") & far["Late P3"]).sum()))
fg = far[far["Play Type"] == "goal"].copy()
fg["Lead of Shooter"] = np.where(fg["Team Making the Play"] == fg["Home Team"],
                                 fg["Cumulated Home Score"] - fg["Cumulated Away Score"],
                                 fg["Cumulated Away Score"] - fg["Cumulated Home Score"])
print("Для далёких голов счёт на момент гола (с учётом этого гола, +: бросавший ведёт):")
print(fg["Lead of Shooter"].describe().round(2).to_string())
print("Доля далёких бросков (>100 ф) среди всех точных, %%: %.2f; среди промахов, %%: %.2f" %
      (100 * (sh[sh["Shot"] == "on goal"]["Shot Length"] > 100).mean(), 100 * (sh[sh["Shot"] == "missed"]["Shot Length"] > 100).mean()))
```

### Пункт 10

```python
header(10, "Доля точных бросков среди незаблокированных")
sh["On Target"] = sh["Play Type"].isin(ON_TARGET)
numshots = sh.groupby(["Game ID", "Team Making the Play"])["On Target"].mean().to_frame("Share of Shots on Goal")
print(numshots["Share of Shots on Goal"])
print(numshots["Share of Shots on Goal"].describe())
```

### Пункт 11

```python
header(11, "Команды с самой высокой и самой низкой долей точных бросков")
avgnumshots = numshots["Share of Shots on Goal"].groupby("Team Making the Play").mean()
print(avgnumshots.nlargest(5).round(3))
print(avgnumshots.nsmallest(5).round(3))
print("Среднее по всем командам: %.3f" % avgnumshots.mean())
```

### Пункт 12

```python
header(12, "Среднее число бросков в створ и голов за игру")
tg = pd.concat([
    games.reset_index()[["Game ID", "Home Team"]].rename(columns={"Home Team": "Team Making the Play"}),
    games.reset_index()[["Game ID", "Away Team"]].rename(columns={"Away Team": "Team Making the Play"}),
])
per_game = pd.DataFrame({
    "Shot": sh[sh["On Target"]].groupby(["Game ID", "Team Making the Play"]).size(),
    "Goal": sh[sh["Play Type"] == "goal"].groupby(["Game ID", "Team Making the Play"]).size(),
})
per_game = tg.set_index(["Game ID", "Team Making the Play"]).join(per_game).fillna(0)   # игры без голов -> 0
avgshots = per_game.groupby("Team Making the Play").mean()
print(avgshots.round(3))
print("Число игр у команды:\n", per_game.groupby("Team Making the Play").size().to_string())
```

### Пункт 13

```python
header(13, "Регрессионная диаграмма: броски в створ и голы")
plt.figure(figsize=(7, 5))
sns.regplot(x="Shot", y="Goal", data=avgshots)
resid = avgshots["Goal"] - np.poly1d(np.polyfit(avgshots["Shot"], avgshots["Goal"], 1))(avgshots["Shot"])
for team in resid.abs().nlargest(6).index:
    plt.annotate(team, (avgshots.loc[team, "Shot"], avgshots.loc[team, "Goal"]),
                 xytext=(4, 4), textcoords="offset points", fontsize=8)
plt.xlabel("Average number of shots on goal per game")
plt.ylabel("Average number of goals per game")
plt.tight_layout()
plt.savefig("figures/fig3_shots_vs_goals.png", dpi=150)
plt.show()
lr = stats.linregress(avgshots["Shot"], avgshots["Goal"])
print("Коэффициент корреляции Пирсона r = %.3f, p-value = %.3f" % (lr.rvalue, lr.pvalue))
print("Наклон = %.3f гола на 1 бросок, R^2 = %.3f" % (lr.slope, lr.rvalue ** 2))
rho = stats.spearmanr(avgshots["Shot"], avgshots["Goal"])
print("Корреляция Спирмена = %.3f, p-value = %.3f" % (rho.statistic, rho.pvalue))
print("Типичная доля реализации (голы/броски в створ): %.3f" % (avgshots["Goal"].sum() / avgshots["Shot"].sum()))
```

### Пункт 15

```python
header(15, "Удары в найденном матче")
m = events[(events["Game ID"] == idxmax) & events["Play Type"].isin(ALL_ATTEMPTS)].copy()
x = m["X Coordinate on the Ice Rink"].copy()
y = m["Y Coordinate on the Ice Rink"].copy()
# Приводим к единой картинке: гости бросают в левые ворота, хозяева — в правые.
# По условию это (x, y) -> (-x, -y) в нечётных периодах; но фактически сторона обороны хозяев
# в данных не всегда 'left' в нечётных периодах, поэтому ориентируемся на поле Home Team's Defending Side.
is_home = m["Team Making the Play"] == m["Home Team"]
needs_flip = (is_home & (m["Home Team's Defending Side"] == "right")) | \
             (~is_home & (m["Home Team's Defending Side"] == "right"))
odd = m["Period"] % 2 == 1
print("Совпадает ли правило «переворот в нечётных периодах» с правилом по полю Defending Side в этом матче:",
      bool((odd == needs_flip).all()))
m["x"] = np.where(needs_flip, -x, x)
m["y"] = np.where(needs_flip, -y, y)
m["Outcome"] = m["Play Type"]
m["Team"] = m["Team Making the Play"]
print("Контроль: средняя x у бросков хозяев = %.1f (ожидаем > 0), у гостей = %.1f (ожидаем < 0)" %
      (m.loc[is_home, "x"].mean(), m.loc[~is_home, "x"].mean()))
plt.figure(figsize=(9, 4.2))
plot_rink(False)
sns.scatterplot(data=m, x="x", y="y", hue="Outcome", style="Team", s=45,
                hue_order=["goal", "shot-on-goal", "blocked-shot", "missed-shot"],
                palette={"goal": "red", "shot-on-goal": "green", "blocked-shot": "gray", "missed-shot": "orange"})
plt.axis("equal")
plt.axis("off")
plt.legend(loc="center left", bbox_to_anchor=(1.0, 0.5), fontsize=8)
plt.title(gameInfo(idxmax), fontsize=9)
plt.tight_layout(pad=0.1)
plt.savefig("figures/fig4_shots_map.png", dpi=150)
plt.show()
print(m.groupby(["Team", "Outcome"]).size().unstack(fill_value=0))
```

### Пункт 19-20

```python
# игры игрока: игра, в которой фамилия встречалась хотя бы в одном поле «на льду»
long = events[["Game ID"] + pl_cols].melt(id_vars="Game ID", value_name="Player").dropna()
gp = long.drop_duplicates(["Game ID", "Player"]).groupby("Player").size().rename("Games")

header(19, "Больше всего бросков в створ за игру")
sog = events[events["Play Type"].isin(ON_TARGET)]["Player in the Event #1"].value_counts().rename("Shots")
t19 = pd.concat([sog, gp], axis=1).dropna()
t19["Shots per Game"] = t19["Shots"] / t19["Games"]
print("Без ограничения на число игр:")
print(t19.sort_values("Shots per Game", ascending=False).head(10).round(3))
print("\nТолько игроки, сыгравшие не менее 5 игр:")
print(t19[t19["Games"] >= 5].sort_values("Shots per Game", ascending=False).head(10).round(3))

header(20, "Больше всего силовых приёмов за игру")
hits = events[events["Play Type"] == "hit"]["Player in the Event #1"].value_counts().rename("Hits")
t20 = pd.concat([hits, gp], axis=1).dropna()
t20["Hits per Game"] = t20["Hits"] / t20["Games"]
print("Без ограничения на число игр:")
print(t20.sort_values("Hits per Game", ascending=False).head(10).round(3))
print("\nТолько игроки, сыгравшие не менее 5 игр:")
print(t20[t20["Games"] >= 5].sort_values("Hits per Game", ascending=False).head(10).round(3))
```

### Пункт 21

```python
header(21, "Вратари, не пропускавшие в матче")
gl = pd.concat([
    pd.DataFrame({"Game ID": events["Game ID"], "Goalie": events["Away Team's Goalie"], "Team": events["Away Team"]}),
    pd.DataFrame({"Game ID": events["Game ID"], "Goalie": events["Home Team's Goalie"], "Team": events["Home Team"]}),
]).dropna().drop_duplicates()                                      # вратарь побывал на льду в игре
# гол пропущен вратарём, если в момент гола на льду стоял вратарь команды, которой забили
g_ev = events[events["Play Type"] == "goal"].copy()
conceded_by_home = g_ev["Team Making the Play"] == g_ev["Away Team"]       # забили гости -> пропустили хозяева
g_ev["Goalie"] = np.where(conceded_by_home, g_ev["Home Team's Goalie"], g_ev["Away Team's Goalie"])
ga = g_ev.dropna(subset=["Goalie"]).groupby(["Game ID", "Goalie"]).size().rename("GA")
gl = gl.merge(ga.reset_index(), on=["Game ID", "Goalie"], how="left").fillna({"GA": 0})
so = gl[gl["GA"] == 0]
print("Голов в пустые ворота (вратаря нет на льду):", int(g_ev["Goalie"].isna().sum()))
res21 = so.groupby(["Goalie", "Team"]).size().rename("Shutout games").reset_index() \
          .sort_values(["Shutout games", "Goalie"], ascending=[False, True])
print(res21.to_string(index=False))
print("Всего вратарей:", len(res21))
# «чистые» матчи (вратарь отыграл всю игру) отдельно
full = events.groupby("Game ID").size()
print("\nДля справки: матчи, где вратарь не пропустил и команда не пропустила вообще (счёт 0 у соперника):")
for gid, r in games.iterrows():
    if r["Away Score"] == 0 or r["Home Score"] == 0:
        print("  ", gameInfo(gid))
```

### Пункт 22

```python
header(22, "Свой вопрос: важен ли первый гол и есть ли преимущество домашнего льда?")
first = events[events["Play Type"] == "goal"].sort_values(["Game ID", "Period", "Play ID"]).groupby("Game ID").head(1).set_index("Game ID")
games["Winner"] = np.where(games["Home Score"] > games["Away Score"], games["Home Team"], games["Away Team"])
games["First Scorer"] = first["Team Making the Play"]
games["First Goal Wins"] = games["First Scorer"] == games["Winner"]
n = games["First Goal Wins"].notna().sum()
k = int(games["First Goal Wins"].sum())
bt = stats.binomtest(k, n, 0.5)
ci = bt.proportion_ci(confidence_level=0.95)
print("Игр с голами: %d; победила команда, забившая первой: %d (%.1f%%), 95%% ДИ [%.2f, %.2f], p = %.4f" %
      (n, k, 100 * k / n, ci.low, ci.high, bt.pvalue))
games["Home Win"] = games["Home Score"] > games["Away Score"]
kh = int(games["Home Win"].sum())
bh = stats.binomtest(kh, len(games), 0.5)
cih = bh.proportion_ci(confidence_level=0.95)
print("Победили хозяева: %d из %d (%.1f%%), 95%% ДИ [%.2f, %.2f], p = %.4f" %
      (kh, len(games), 100 * kh / len(games), cih.low, cih.high, bh.pvalue))
games["Home First"] = games["First Scorer"] == games["Home Team"]
print("Хозяева забивали первыми в %.1f%% игр" % (100 * games["Home First"].mean()))
print("Побед команды, пропустившей первой (то есть отыгравшейся): %d" % int((~games["First Goal Wins"]).sum()))
# вероятность гола в зависимости от расстояния
sh["Dist bin"] = pd.cut(sh["Shot Length"], [0, 10, 20, 30, 40, 60, 100, 200], include_lowest=True)
conv = sh[sh["Play Type"] != "missed-shot"].groupby("Dist bin", observed=True).apply(lambda d: (d["Play Type"] == "goal").mean())
print("\nДоля голов среди бросков в створ по расстоянию:\n", conv.round(3).to_string())
fig, ax = plt.subplots(1, 2, figsize=(10, 3.8))
sns.barplot(x=["Первая забила\nпобедила", "Первая забила\nпроиграла"], y=[k / n, 1 - k / n], ax=ax[0], color="steelblue")
ax[0].axhline(0.5, color="k", ls="--")
ax[0].set_ylabel("Доля игр")
ax[0].set_title("Первый гол и исход игры")
sns.barplot(x=conv.index.astype(str), y=conv.values, ax=ax[1], color="indianred")
ax[1].set_xlabel("Расстояние до ворот, футы")
ax[1].set_ylabel("Доля голов среди бросков в створ")
ax[1].set_title("Реализация бросков в створ")
ax[1].tick_params(axis="x", rotation=30)
plt.tight_layout()
plt.savefig("figures/fig5_own_question.png", dpi=150)
plt.show()
```
